```python
from diffusers import DiffusionPipeline

pipe = DiffusionPipeline.from_pretrained("stable-diffusion-v1-5/stable-diffusion-v1-5")

prompt = "Astronaut in a jungle, cold color palette, muted colors, detailed"
image = pipe(prompt).images[0]
```

# Image to Caption

BLIP (Bootstrapped Language-Image Pretraining) is specifically designed for image-to-text tasks like captioning, visual question answering, and explainability.

It generates natural language captions that are descriptive and coherent.

The base model is fast and lightweight for testing, while the large model is more accurate.

```python
!pip install transformers torch torchvision pillow

from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image

# Assuming 'image' is your PIL image from stable diffusion
# If not, convert it: image = Image.open("path_to_image.jpg")

# Load BLIP model and processor
processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")

# Preprocess and generate caption
inputs = processor(images=image, return_tensors="pt")
out = model.generate(**inputs)
caption = processor.decode(out[0], skip_special_tokens=True)

print("Caption:", caption)
```

## Result
Caption: a man in an astronaut suit standing in the jungle

The result caption of image to text is correct.
